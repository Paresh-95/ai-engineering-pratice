```python
#Exception Handling 

class APIValidationError(Exception):
    """Custom exception raised when API Validations fails"""
    pass 

def process_api_payload(payload:dict):
    if "prompt" not in payload:
        raise APIValidationError("Missing required 'prompt' key in payload")

    return f"Processing prompt: {payload['prompt']}"


#Full Exception Handling Lifecycle 


def execute_pipeline(payload:dict):
    try:
        print("Starting the pipeline")
        result = process_api_payload(payload)
    except APIValidationError as err:
        print(f'Caught specific API Validation error: {err}')
    except Exception as err:
        print(f'Caught specific generic error: {err}')
    else:
        print(f'Pipeline succeed: {result}')
    finally:
        print(f'Cleanup: closing connection and other links')


print("----Extracting with invalid paylload")
execute_pipeline({})

# print("----Extracting with invalid paylload")
# execute_pipeline({"prompt":"Hi"})
```

```python
#Homework Questions  @1

def demo(raw_price,raw_quantity,raw_discount):
    total_cost = float(raw_price)*int(raw_quantity)

    
    if (float(raw_discount)):
        discount_price = total_cost - (total_cost * float(raw_discount) / 100)
    else:
        discount_price = total_cost


    if total_cost is None:
        print(f"Total cost is None -> {total_cost}")

    if not total_cost:
        print(f"Total cost is falsy (for example it could be 0 )")

    return discount_price


result = demo("199.99","5","10")
print(f"final price {result}")
```

```python
# Homework Q3

responses = [500,502,503,200,404]
attempt = 0
waitTime = 1

for response in responses:

    if response is None or not isinstance(response,int):
        continue

    if response == 500 or response ==502 or response==503:
        attempt+=1

        print(f"Warning: Recieved {response}. Retrying...")
        print(f"Attempt: {attempt}, Wait time: {waitTime} seconds")

        waitTime *=2
    elif response ==200:
        print(f"success {response}: Api request completed")
        break
    else:
        print(f"fatal error: Recieved {response} Stopping.")
        break

    
```

```python
# Homework Q4

chat_logs = [
    {"user": "Alice", "tags": ["python", "ai"], "msg": "Hello world"},
    {"user": "Bob", "tags": ["rag", "ai"], "msg": "Building a RAG pipeline"},
    {"user": "Alice", "tags": ["python"], "msg": "Learning collections"},
]

# Set to store the uniques tags
unique_tags  = set()

#Dictionary to store user and there msgs
user_index = {}


for chat in chat_logs:

    #Add all tags in the set 
    for tag in chat["tags"]:
        unique_tags.add(tag)

    user = chat["user"]
    message = chat["msg"]

    # Get existing messages for the user safely
    messages = user_index.get(user,[])

    # add the current message 
    messages.append(message)

    #Store the msg back in the dict
    user_index[user] = messages


print("Unique Tags Set:")
print(unique_tags)

print("User Index Dict:")
print(user_index)

```

```python
# Homework Q5
from datetime import datetime


def add_log_entry(event:str,level:str="INFO",logs:list[str] = None)->list[str]:
    if logs is None:
        logs = []


    #Create a timestamp log msg
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    log_entry = f"[{timestamp}] [{level}] {event}"

    logs.append(log_entry)
    return logs


logs = add_log_entry("RAG pipeline started")
print(logs)

logs = add_log_entry("Embedding generation completed", "INFO", logs)
print(logs)
```
