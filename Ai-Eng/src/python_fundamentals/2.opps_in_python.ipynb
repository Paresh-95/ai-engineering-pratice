```python
# OOPS => Classes and Objects 

class Dog:
    species = "Dog"  #Class Variable

    @classmethod   # class method -> static method (decorator) factory mthd (can be used construct obj in different way )
    def walk(cls): # have access of cls (class)
        print(f"{cls.species} is walking")

    @staticmethod  # static method -> pure static method (mostly utility)
    def run():
        print("Dog is running")

    
    def __init__(self,name:str,breed:str):   #Dunder method (__init__ refers to contrutor self-> this keyword in java refer to current object)
        self.name = name   # Instnce Varibale
        self.breed = breed
        self.eyes = 2

    def bark(self):
        return f"{self.name} is barking Woff Woff! and have {self.eyes} eyes"

leo = Dog(name="leo",breed="Labra")

print(leo.breed)
print(leo.bark())

Dog.walk()
Dog.run()
```

```python
# OOPS =>  class method 

class User:

    def __init__(self,name):
        self.name = name
 
    @classmethod   #factory method
    def from_dict(cls,data):
        return cls(data["name"])


user = User.from_dict({"name":"Paresh"})
print(user.name)
```

```python
# OOPS =>  static method  (pure static methods like java)

class MathUtils:

    @staticmethod
    def add(a,b):
        return a+b

print(MathUtils.add(10,20))
```

```python
# OOPS =>  property

class User:
    def __init__(self,first,last):
        self.first = first
        self.last = last

    @property
    def full_name(self):
        return f"{self.first} {self.last}"

user.full_name
```

```python
# OOPS => Inheritance

class Animal:

    def __init__(self,name:str):
        self.name = name

    def speak(self) -> str:
        return "Some Sound"

class Dog(Animal):
     
    def __init__(self,name:str,breed:str):
            super().__init__(name)
            self.breed = breed
    
    def speak(self) -> str:
        return "Woff Woff"

gen_animal = Animal("Some Animal")
rex = Dog("rex","labrador")

print(f"gen_animal.speak() --> {gen_animal.speak()}")
print(f"rex.speak()   ---->  {rex.speak()}")
print(f"rex.name  (from parent __init__ )    ---> {rex.name}")
print(f"rex.breed (from dogs own __init__)   --->  {rex.breed}")
```

```python
#Variadic Argurments    *args **kwargs

def add_numbers(*args):
    print(type(args))
    print(args)


add_numbers(1,2,3,4,"True")


def print_profile(**kwargs):
    print(kwargs)
    print(type(kwargs))


print_profile(name="Paresh",age=30,is_active=True)




def create_profile(name,role,experience):
    print(name)
    print(role)
    print(experience)

profile = {
    "name":"Paresh",
    "role":"SDE",
    "experience": 14
}

create_profile(name="Paresh",role="SDE",experience=14)
create_profile(**profile)



def add(a,b,c):
    return sum((a,b,c))

k = [1,2,3]

print(add(*k));

```
