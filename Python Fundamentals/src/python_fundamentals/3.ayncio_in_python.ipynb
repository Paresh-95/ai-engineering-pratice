```python
import asyncio
import time

async def task_a():
    print("Task A Started")
    await asyncio.sleep(2)
    print("Task A Ended")


async def task_b():
    print("Task B Started")
    await asyncio.sleep(1)
    print("Task B Ended")

async def main():
    await task_a()
    await task_b()


start = time.perf_counter()


await main()

print(f"Time Taken: {time.perf_counter()-start:.2f} seconds")
```

```python

import asyncio
import time

async def task_a():
    print("Task A Started")
    await asyncio.sleep(2)
    print("Task A Ended")


async def task_b():
    print("Task B Started")
    await asyncio.sleep(1)
    print("Task B Ended")

async def main():
    result = await asyncio.gather(
        task_a(),
        task_b()
    )
    


start = time.perf_counter()


await main()

print(f"Time Taken: {time.perf_counter()-start:.2f} seconds")
```

```python
import asyncio


async def generate_numbers():


    for number in range(5):
        await asyncio.sleep(1)
        yield number

async def main():

    async for number in generate_numbers():
        print(number)

await main()
```

```python
import asyncio
import httpx


async def fetch_post():
    async with httpx.AsyncClient() as client:

        response  = await client.get(
            "https://jsonplaceholder.typicode.com/posts/1"
        )

        return response.json()


async def main():
    post = await fetch_post()
    print(post)



await main()
```

```python
import asyncio
import httpx

async def fetch_user(client):
    response = await client.get("https://jsonplaceholder.typicode.com/users/1")
    return response.json()

async def fetch_post(client):
    response = await client.get("https://jsonplaceholder.typicode.com/posts/1")
    return response.json()

async def fetch_todo(client):
    response = await client.get("https://jsonplaceholder.typicode.com/todos/1")
    return response.json()



async def main():
    async with httpx.AsyncClient() as client:
            user, post, todo = await asyncio.gather(
            fetch_user(client),
            fetch_post(client),
            fetch_todo(client)
        )


    print("\nUsers: ")
    print(user["name"])

    print("\nPosts: ")
    print(post["title"])

    print("\nTodos: ")
    print(todo["title"])


await main()
```
