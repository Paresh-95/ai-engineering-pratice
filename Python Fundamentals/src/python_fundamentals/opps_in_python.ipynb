```python
# OOPS => Classes and Objects 

class Dog:
    species = "Dog"  #Class Variable

    @classmethod   # class method -> static method (decorator) factory mthd (can be used construct obj in different way )
    def walk(cls): # have access of cls (class)
        print(f"{cls.species} is walking")

    @staticmethod  # static method -> pure static method (mostly utility)
    def run():
        print("Dog is running")

    
    def __init__(self,name:str,breed:str):   #Dunder method (__init__ refers to contrutor self-> this keyword in java refer to current object)
        self.name = name   # Instnce Varibale
        self.breed = breed
        self.eyes = 2

    def bark(self):
        return f"{self.name} is barking Woff Woff! and have {self.eyes} eyes"

leo = Dog(name="leo",breed="Labra")

print(leo.breed)
print(leo.bark())

Dog.walk()
Dog.run()
```

```python
# OOPS =>  class method 

class User:

    def __init__(self,name):
        self.name = name
 
    @classmethod   #factory method
    def from_dict(cls,data):
        return cls(data["name"])


user = User.from_dict({"name":"Paresh"})
print(user.name)
```

```python
# OOPS =>  static method  (pure static methods like java)

class MathUtils:

    @staticmethod
    def add(a,b):
        return a+b

print(MathUtils.add(10,20))
```

```python
# OOPS =>  property

class User:
    def __init__(self,first,last):
        self.first = first
        self.last = last

    @property
    def full_name(self):
        return f"{self.first} {self.last}"

user.full_name
```
